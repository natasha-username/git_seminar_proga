```python
import spacy
from spacy.lang.ru.examples import sentences

nlp = spacy.load("ru_core_news_sm")
```

```python
def text_preprocess(text):
  a = []
  text = re.sub(r'\r?\n', ' ', text)
  text = re.sub(r'-', ' ', text)
  text = re.sub(r'[^\w\s]', '', text)
  text = re.sub(r'  ', '', text)
  text1 = text.lower()
  text2 = text1.strip()
  for word in text2:
    if word in sw:
      text2.replace(word, "")
  doc = nlp(text2)
  for token in doc:
    lemma = token.lemma_
    a.append(lemma)
    b = ' '.join(a)
  return(b)
```

```python
b = ''''
Такой зеленый, серый, то есть
весь заштрихованный дождем,
и липовое, столь густое,
что я перенести - уйдем!'''
text_preprocess(b)
```
